## 1. Validação por função booleana (cap. 3)

```python
def leitura_valida(valor, minimo=0, maximo=100):
    """Retorna True/False. Usa parametros DEFAULT (cap. 3)."""
    return valor >= minimo and valor <= maximo


def energia_segura(valor):
    return leitura_valida(valor, 0, 200)
```

## 2. Análise de energia

```python
def analisar_energia(geracao, consumo, reserva=0):
    """if/elif/else classico (cap. 2). reserva e parametro default (cap. 3)."""
    saldo = geracao - consumo
    if consumo > geracao:
        situacao, mensagem = "RISCO", "ALERTA: consumo maior que geracao"
    elif geracao > consumo:
        situacao, mensagem = "SOBRA", "SUGESTAO: armazenar energia excedente"
    else:
        situacao, mensagem = "EQUILIBRIO", "OK: geracao e consumo equilibrados"
    return {"geracao": geracao, "consumo": consumo, "reserva": reserva,
            "saldo": saldo, "situacao": situacao, "mensagem": mensagem}
```

## 3. Decisão principal — lógica booleana combinada

Combina múltiplos critérios em ordem de prioridade:
1. **Crítico** = energia<30 **AND** consumo≥60 → emergência
2. **Alto** = tempestade **AND** energia<50 → economia
3. **Médio** = energia<50 → reduzir consumo
4. **Normal** = caso contrário

```python
def decidir_acao(energia, consumo, previsao_tempestade=False):
    consumo_alto    = consumo >= 60
    energia_critica = energia < 30
    energia_baixa   = energia < 50

    if energia_critica and consumo_alto:
        return {"nivel": "CRITICO", "acao": "ATIVAR MODO DE EMERGENCIA",
                "detalhe": "Energia critica com consumo alto."}
    elif previsao_tempestade and energia_baixa:
        return {"nivel": "ALTO", "acao": "ATIVAR MODO DE ECONOMIA",
                "detalhe": "Tempestade prevista e energia baixa."}
    elif energia_baixa:
        return {"nivel": "MEDIO", "acao": "REDUZIR CONSUMO",
                "detalhe": "Energia abaixo do nivel seguro (50)."}
    else:
        return {"nivel": "NORMAL", "acao": "MANTER SISTEMAS NORMAIS",
                "detalhe": "Condicoes dentro do esperado."}
```

## 4. Priorização de módulos

```python
def priorizar_modulos(modulos, modo_economia):
    ligados, desligados, consumo_final = [], [], 0
    for nome in modulos:
        info = modulos[nome]
        if info["essencial"]:
            ligados.append(nome)
            consumo_final += info["consumo"]
        elif modo_economia:
            desligados.append(nome)
        else:
            ligados.append(nome)
            consumo_final += info["consumo"]
    return {"ligados": ligados, "desligados": desligados,
            "consumo_final": consumo_final}
```

## 5. Recursividade (cap. 3) — propagação de alertas

Modelo: `contagemRecursiva()` do capítulo. Tem **condição de parada**
(índice fora do range) e **argumento que evolui** (índice + 1).

```python
def propagar_alerta(setores, indice=0, mensagem="ALERTA"):
    if indice >= len(setores):                  # condicao de parada
        return []
    cabeca = f"[{mensagem}] -> setor '{setores[indice]}'"
    cauda = propagar_alerta(setores, indice + 1, mensagem)
    return [cabeca] + cauda
```

## 6. Demonstração — quatro cenários

```python
cenarios = [
    ("Energia critica + consumo alto",       25, 70, False),
    ("Tempestade prevista + energia baixa",  45, 40, True),
    ("Energia baixa, sem tempestade",        48, 35, False),
    ("Condicoes normais",                    90, 50, False),
]
for nome, e, c, t in cenarios:
    d = decidir_acao(e, c, t)
    print(f"- {nome}")
    print(f"  Entrada: energia={e}, consumo={c}, tempestade={t}")
    print(f"  Saida  : [{d['nivel']}] {d['acao']}")
    print()

print("Recursividade (cap. 3):")
alertas = propagar_alerta(["habitat-1", "estufa", "lab", "energia"],
                          mensagem="ALERTA O2 BAIXO")
for a in alertas:
    print("  " + a)
```
