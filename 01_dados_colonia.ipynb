# Organização dos Dados — Colônia Aurora Siger

**Capítulos aplicados:** 4 (estruturas avançadas) e 5 (vetores como listas).

Este notebook organiza os dados em:
- **Hierarquia** (árvore não-binária via dicionário aninhado);
- **Tabela hash** (o `dict` do Python tem acesso médio O(1));
- **Listas** que funcionam como vetores 1D (séries históricas).

Sem bibliotecas externas — apenas Python puro.

## 1. Estrutura hierárquica da colônia

```python
def criar_colonia():
    """Monta a colonia como hierarquia (arvore nao-binaria)."""
    return {
        "energetico": {
            "solar": {
                "potencia_atual": 45,
                "historico_irradiancia": [120, 180, 240, 300, 360, 420, 480],
                "historico_geracao":     [12,  19,  26,  33,  40,  47,  54],
                "area_paineis_m2": 20,
                "eficiencia": 0.18,
            },
            "eolico": {
                "potencia_atual": 25,
                "historico_vento":   [8, 10, 12, 13, 15],
                "historico_geracao": [18, 21, 25, 27, 31],
                "area_rotor_m2": 12,
                "rho_ar": 0.020,
                "cut_in": 3, "v_nominal": 12, "cut_out": 25,
                "cp": 0.40,
            },
            "reserva_baterias": 60,
            "capacidade_baterias": 100,
        },
        "ambiental": {
            "temperatura_interna": 22,
            "temperatura_externa": -55,
            "velocidade_vento": 14,
            "irradiancia_atual": 360,
            "previsao_tempestade": False,
        },
        "operacional": {
            "consumo_total": 70,
            "modulos": {
                "suporte_vida": {"consumo": 30, "essencial": True},
                "comunicacao":  {"consumo": 8,  "essencial": True},
                "estufa":       {"consumo": 20, "essencial": False},
                "laboratorio":  {"consumo": 12, "essencial": False},
            },
        },
        "sensores": {
            "S001": {"tipo": "temperatura", "valor": 22,  "ok": True},
            "S002": {"tipo": "pressao",     "valor": 101, "ok": True},
            "S003": {"tipo": "oxigenio",    "valor": 21,  "ok": True},
            "S004": {"tipo": "umidade",     "valor": 45,  "ok": True},
        },
    }
```

## 2. Navegação na hierarquia

A função `acessar(colonia, caminho)` percorre a árvore usando uma
lista de chaves como caminho.

```python
def acessar(colonia, caminho):
    """Percorre a hierarquia usando uma lista de chaves."""
    atual = colonia
    for chave in caminho:
        if isinstance(atual, dict) and chave in atual:
            atual = atual[chave]
        else:
            return None
    return atual
```

## 3. Tabela hash didática (cap. 4)

Função hash simples: `h(k) = soma_codigos(k) mod m`. Para inteiros,
é apenas `k mod m`. O `dict` do Python já faz isso internamente.

```python
def funcao_hash(chave, tamanho):
    """h(k) = (soma dos codigos dos caracteres) mod tamanho."""
    if isinstance(chave, int):
        return chave % tamanho
    soma = 0
    for c in str(chave):
        soma += ord(c)
    return soma % tamanho


def buscar_sensor(colonia, id_sensor):
    """Acesso O(1) ao sensor pelo ID (cap. 4)."""
    return colonia["sensores"].get(id_sensor)
```

## 4. Funções auxiliares

```python
def energia_total_gerada(colonia):
    solar = acessar(colonia, ["energetico", "solar", "potencia_atual"])
    eolico = acessar(colonia, ["energetico", "eolico", "potencia_atual"])
    return solar + eolico


def listar_modulos_nao_essenciais(colonia):
    modulos = acessar(colonia, ["operacional", "modulos"])
    return [n for n in modulos if not modulos[n]["essencial"]]


def resumo(colonia):
    print("=" * 52)
    print("ESTADO DA COLONIA AURORA SIGER")
    print("=" * 52)
    print("Geracao solar .......:", acessar(colonia, ["energetico","solar","potencia_atual"]))
    print("Geracao eolica ......:", acessar(colonia, ["energetico","eolico","potencia_atual"]))
    print("Geracao total .......:", energia_total_gerada(colonia))
    print("Consumo total .......:", acessar(colonia, ["operacional","consumo_total"]))
    print("=" * 52)
```

## 5. Demonstração

```python
colonia = criar_colonia()
resumo(colonia)

print()
print("Caminho na hierarquia (arvore):")
print("  geracao solar  =", acessar(colonia, ["energetico", "solar", "potencia_atual"]))
print("  vento atual    =", acessar(colonia, ["ambiental", "velocidade_vento"]))

print()
print("Acesso por hash (cap. 4):")
print("  Sensor S001 =", buscar_sensor(colonia, "S001"))
print("  Sensor S999 =", buscar_sensor(colonia, "S999"))
print("  funcao_hash('S001', 7) =", funcao_hash("S001", 7))

print()
print("Modulos nao essenciais:", listar_modulos_nao_essenciais(colonia))
```
